# Python Variables

A **variable** is a name that refers to a value stored in memory. Think of it as a **label on a box**.

```
name = "Sara"
```
`=` is the **assignment operator**. It does *not* mean "equals" in the math sense.

````python
#@title { display-mode: "form" }
from IPython.display import HTML, display
import uuid

def refviz(names, objs, before, after=None, cap1="", cap2="", delay=3):
    """Name tags (left) pointing to objects (right). before/after: {name: object_index}."""
    after = after or before
    H = 30 + max(len(names), len(objs)) * 70
    ty = {n: 20 + i * 70 for i, n in enumerate(names)}
    oy = [20 + j * 70 for j in range(len(objs))]
    mid = uuid.uuid4().hex[:6]
    used_b, used_a = set(before.values()), set(after.values())
    s = [f'<svg viewBox="0 0 560 {H}" width="100%" style="max-width:560px"><defs>'
         f'<marker id="m{mid}" markerWidth="10" markerHeight="8" refX="9" refY="4" orient="auto">'
         f'<path d="M0,0 L10,4 L0,8 z" fill="#facc15"/></marker></defs>']
    for j, (label, typ) in enumerate(objs):
        if j in used_b and j in used_a: anim = ""
        elif j in used_a: anim = f'style="opacity:0;animation:fin .6s {delay}s forwards"'
        elif j in used_b: anim = f'style="animation:dim .6s {delay + .8}s forwards"'
        else: anim = 'style="opacity:0"'
        s.append(f'<g {anim}><rect x="330" y="{oy[j]}" width="210" height="50" rx="10" fill="#1e293b" stroke="#475569"/>'
                 f'<text x="435" y="{oy[j] + 24}" fill="#f8fafc" font-size="17" text-anchor="middle" font-family="monospace">{label}</text>'
                 f'<text x="435" y="{oy[j] + 42}" fill="#94a3b8" font-size="12" text-anchor="middle">{typ}</text></g>')
    for n in names:
        s.append(f'<rect x="30" y="{ty[n]}" width="80" height="50" rx="25" fill="#38bdf8"/>'
                 f'<text x="70" y="{ty[n] + 32}" fill="#0f172a" font-size="22" font-weight="bold" text-anchor="middle" font-family="monospace">{n}</text>')
    def line(n, j, style):
        return (f'<line x1="112" y1="{ty[n] + 25}" x2="326" y2="{oy[j] + 25}" stroke="#facc15" '
                f'stroke-width="3" marker-end="url(#m{mid})" {style}/>')
    for n in names:
        b, a = before[n], after[n]
        if b == a:
            s.append(line(n, b, ""))
        else:
            s.append(line(n, b, f'style="animation:fout .5s {delay}s forwards"'))
            s.append(line(n, a, f'style="opacity:0;animation:fin .6s {delay + .3}s forwards"'))
    s.append('</svg>')
    if cap2:
        cap = (f'<div style="display:grid"><div style="grid-area:1/1;animation:fout .5s {delay}s forwards">{cap1}</div>'
               f'<div style="grid-area:1/1;opacity:0;animation:fin .6s {delay + .3}s forwards">{cap2}</div></div>')
    else:
        cap = f'<div>{cap1}</div>'
    return HTML(f'''<style>@keyframes fin{{to{{opacity:1}}}} @keyframes fout{{to{{opacity:0}}}} @keyframes dim{{to{{opacity:.25}}}}</style>
<div style="background:#0f172a;padding:16px;border-radius:12px;color:#e2e8f0;font:17px monospace">{"".join(s)}
<div style="margin-top:8px">{cap}</div></div>''')

refviz(
    names=["name"],
    objs=[('"Sara"', "str")],
    before={"name": 0},
    cap1='<b>name = "Sara"</b>: Python creates a str object and sticks the label <b>name</b> on it',
)

````

````python
# @title Reassigning the same variable
refviz(
    names=["name"],
    objs=[('"Sara"', "str"), ('"Saad"', "str")],
    before={"name": 0},
    after={"name": 1},
    cap1='<b>name = "Sara"</b>: the label points to "Sara"',
    cap2='<b>name = "Saad"</b>: a new box is created and the label moves. "Sara" has no label left, so Python will clean it up',
    delay=3,
)
````

````python
# @title Two variables, then one is reassigned
refviz(
    names=["name", "other"],
    objs=[('"Sara"', "str"), ('"Saad"', "str")],
    before={"name": 0, "other": 0},
    after={"name": 1, "other": 0},
    cap1='<b>other = name</b>: both labels are on the same "Sara" box',
    cap2='<b>name = "Saad"</b>: only <b>name</b> moves. <b>other</b> still points to "Sara", so that box stays',
    delay=3,
)
````

Invalid: `2name`, `my-var`, `my var`, `class`

### Dynamic typing: a variable can be re-pointed to any type

````python
x = 10
print(x, type(x))

x = 3.14
print(x, type(x))

x = 2 + 3j
print(x, type(x))

x = "now I'm a string"
print(x, type(x))

x = True
print(x, type(x))

x = None
print(x, type(x))
````

### Multiple assignment & swapping

````python
a, b, c = 1, 2, 3            # unpack three values
print(a, b, c)

x = y = z = 0                # same value to all
print(x, y, z)

# Swap two variables (classic Python trick, no temp variable needed)
a, b = b, a
print(a, b)
````

### Variables are references

````python
p = 100
q = p
print(id(p) == id(q))   # same object in memory

q = 200
print(p, q)             # p unchanged: integers are immutable
````

````python
#@title { display-mode: "form" }
import ipywidgets as w
from IPython.display import HTML, display

i100, i200 = hex(id(100)), hex(id(200))

def tag(n, y):
    return (f'<rect x="30" y="{y}" width="80" height="50" rx="25" fill="#38bdf8"/>'
            f'<text x="70" y="{y+32}" fill="#0f172a" font-size="22" font-weight="bold" text-anchor="middle">{n}</text>')
def obj(v, i, y):
    return (f'<rect x="330" y="{y}" width="210" height="64" rx="10" fill="#1e293b" stroke="#475569"/>'
            f'<text x="435" y="{y+26}" fill="#f8fafc" font-size="20" text-anchor="middle">{v}</text>'
            f'<text x="435" y="{y+48}" fill="#94a3b8" font-size="12" text-anchor="middle">int · id {i}</text>')
def arr(y1, y2):
    return f'<line x1="112" y1="{y1}" x2="326" y2="{y2}" stroke="#facc15" stroke-width="3" marker-end="url(#arw)"/>'

caps = ["Click Next ▶ to run the code line by line",
        "<b>p = 100</b> → the name p points to an int object",
        "<b>q = p</b> → q points to the SAME object. Nothing is copied!",
        "<b>q = 200</b> → q moves to a NEW object. p is untouched."]
badges = ["", "",
          '<span style="color:#4ade80">id(p) == id(q) → True</span>',
          '<span style="color:#f87171">id(p) == id(q) → False</span><br>print(p, q) → 100 200']

def render(step):
    svg = ('<svg viewBox="0 0 560 220" width="100%" style="max-width:560px"><defs>'
           '<marker id="arw" markerWidth="10" markerHeight="8" refX="9" refY="4" orient="auto">'
           '<path d="M0,0 L10,4 L0,8 z" fill="#facc15"/></marker></defs>'
           + (obj(100, i100, 20) + tag("p", 27) + arr(52, 52) if step >= 1 else "")
           + (tag("q", 137) + (arr(162, 64) if step == 2 else "") if step >= 2 else "")
           + (obj(200, i200, 130) + arr(162, 162) if step >= 3 else "")
           + '</svg>')
    return HTML(f'''<div style="background:#0f172a;padding:16px;border-radius:12px;color:#e2e8f0;font:17px monospace">
{svg}<div style="min-height:30px;margin-top:6px">{caps[step]}</div>
<div style="min-height:52px;margin-top:8px">{badges[step]}</div></div>''')

state = {"step": 0}
out = w.Output()
nxt = w.Button(description="▶", button_style="primary")
rst = w.Button(description="↺")

def show():
    with out:
        out.clear_output(wait=True)
        display(render(state["step"]))

def go(_):
    state["step"] = min(state["step"] + 1, 3); show()
def reset(_):
    state["step"] = 0; show()

nxt.on_click(go); rst.on_click(reset)
display(w.HBox([nxt, rst]), out)
show()
````

# Data Types and the `type()` Function

| Type | Example | Description |
|---|---|---|
| `int` | `42`, `-7` | Whole numbers (unlimited size) |
| `float` | `3.14`, `-0.5`, `2e3` | Decimal numbers |
| `str` | `"hello"` | Text |
| `bool` | `True`, `False` | Logical values |
| `NoneType` | `None` | "No value" |
| `complex` | `3+4j` | Complex numbers (rare) |

Later we'll meet `list`, `tuple`, `dict`, `set`.

````python
#@title { display-mode: "form" }
import ipywidgets as w
from IPython.display import HTML, display

# ---------- stepper: ◀ ▶ ↺ buttons ----------
def stepper(frames):
    """frames: list of HTML strings, one per click."""
    state = {"i": 0}
    out = w.Output()
    back = w.Button(description="◀", layout=w.Layout(width="50px"))
    nxt = w.Button(description="▶", button_style="primary", layout=w.Layout(width="50px"))
    rst = w.Button(description="↺", layout=w.Layout(width="50px"))

    def show():
        with out:
            out.clear_output(wait=True)
            display(HTML(frames[state["i"]]))

    def move(d):
        def f(_):
            state["i"] = max(0, min(len(frames) - 1, state["i"] + d))
            show()
        return f

    back.on_click(move(-1)); nxt.on_click(move(1))
    rst.on_click(lambda _: (state.update(i=0), show()))
    display(w.HBox([back, nxt, rst]), out)
    show()

# ---------- visual kit ----------
COL = {"int": "#38bdf8", "float": "#fb923c", "str": "#4ade80", "bool": "#c084fc", "list": "#f472b6", "class": "#e2e8f0"}

def chip(val, typ, label=None):
    c = COL[typ]
    return (f'<div style="display:inline-block;text-align:center;margin:0 8px">'
            f'<div style="border:2px solid {c};color:{c};border-radius:10px;padding:10px 16px;font-size:22px">{val}</div>'
            f'<div style="color:#94a3b8;font-size:13px;margin-top:4px">{label or typ}</div></div>')

def op(s):
    return f'<span style="display:inline-block;vertical-align:top;margin:12px 6px;font-size:20px;color:#94a3b8">{s}</span>'

def arrow(label):
    return f'<div style="display:inline-block;vertical-align:top;margin:4px 8px;color:#facc15">{label}<br>──────▶</div>'

def panel(body, code, note=""):
    return (f'<div style="background:#0f172a;color:#e2e8f0;font:17px monospace;padding:18px;border-radius:12px;text-align:center">'
            f'<div style="color:#facc15;margin-bottom:14px;text-align:left">{code}</div>{body}'
            f'<div style="margin-top:14px;min-height:24px">{note}</div></div>')
COL.update({"NoneType": "#94a3b8", "complex": "#fde047"})

frames = [
    # ---- one frame per type from your table ----
    panel(chip("42", "int") + chip("-7", "int"),
          "x = 42",
          "<b>int</b>: whole numbers, unlimited size &nbsp; → &nbsp;type(x) = &lt;class 'int'&gt;"),

    panel(chip("3.14", "float") + chip("-0.5", "float") + chip("2e3", "float"),
          "x = 3.14",
          "<b>float</b>: decimal numbers. <b>2e3</b> means 2 × 10³ = 2000.0"),

    panel(chip('"hello"', "str"),
          'x = "hello"',
          "<b>str</b>: text, always inside quotes"),

    panel(chip("True", "bool") + chip("False", "bool"),
          "x = True",
          "<b>bool</b>: logical values. Note the capital T and F"),

    panel(chip("None", "NoneType"),
          "x = None",
          "<b>NoneType</b>: the 'no value' object. There is only one in the whole program"),

    panel(chip("3+4j", "complex"),
          "x = 3 + 4j",
          "<b>complex</b>: real part + imaginary part. Rare outside science and engineering"),

    # ---- the label moves, the type stays with the object ----
    panel(chip("10", "int") + arrow("x = 3.14") + chip("3.14", "float"),
          "x = 10 &nbsp;→&nbsp; x = 3.14",
          "The label <b>x</b> moves to a new box. The <b>type belongs to the box</b>, not the label"),

    # ---- mixing types ----
    panel(chip("1", "int") + op("+") + chip("2.5", "float") + op("=") + chip("3.5", "float"),
          "1 + 2.5",
          "int + float gives a <b>float</b>. Python widens to the more general type"),

    panel(chip("7", "int") + op("/") + chip("2", "int") + op("=") + chip("3.5", "float"),
          "7 / 2",
          "<b>/</b> always gives a float, even for two ints"),

    panel(chip("7", "int") + op("//") + chip("2", "int") + op("=") + chip("3", "int"),
          "7 // 2",
          "<b>//</b> is floor division and keeps an int"),

    panel(chip("True", "bool") + op("+") + chip("True", "bool") + op("=") + chip("2", "int"),
          "True + True",
          "bool behaves like a number: True is 1 and False is 0. The result is an <b>int</b>"),

    # ---- converting types ----
    panel(chip('"5"', "str") + arrow("int()") + chip("5", "int"),
          'int("5")',
          "Converting a <b>str</b> to an <b>int</b> creates a new box. The original is untouched"),

    panel(chip('"5"', "str") + op("+") + chip('"5"', "str") + op("=") + chip('"55"', "str"),
          '"5" + "5"',
          "On <b>str</b>, + joins text. Compare with the next frame"),

    panel(chip("5", "int") + op("+") + chip("5", "int") + op("=") + chip("10", "int"),
          "5 + 5",
          "On <b>int</b>, + adds numbers. Same symbol, different behavior per type"),
]

stepper(frames)
````

# Expressions and Operators

An **expression** is anything that produces a value (`2 + 3`, `age > 18`). **Operators** are the symbols that build expressions.

## 4.1 Arithmetic Operators
| Op | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `7 + 2` | 9 |
| `-` | Subtraction | `7 - 2` | 5 |
| `*` | Multiplication | `7 * 2` | 14 |
| `/` | Division (always float) | `7 / 2` | 3.5 |
| `//` | Floor division | `7 // 2` | 3 |
| `%` | Modulus (remainder) | `7 % 2` | 1 |
| `**` | Exponent | `7 ** 2` | 49 |

````python
a, b = 17, 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
print("-17 // 5 =", -17 // 5, "  (floors toward -infinity!)")
````

**Operator precedence** (high → low): `**` → `*  /  //  %` → `+  -`. Use parentheses to be clear.

````python
print(2 + 3 * 4)        # 14
print((2 + 3) * 4)      # 20
print(2 ** 3 ** 2)      # 512
print(-3 ** 2)          # -9  (** binds tighter than unary minus)
````

## 4.3 Comparison Operators
Always produce a `bool` (`True`/`False`).

| Op | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal |
| `>`  `<` | greater / less than |
| `>=`  `<=` | greater/less or equal |

Note: `=` assigns, `==` compares!

````python
x, y = 10, 20
print(x == y, x != y, x > y, x < y, x >= 10, y <= 15)

# Chained comparisons (unique and handy in Python)
age = 25
print(18 <= age <= 30)

# Comparing strings (alphabetical / by Unicode order)
print("apple" < "aanana")
print("Zebra" < "apple")      # uppercase letters come before lowercase!
print("abc" == "ABC")
````

## 4.4 Logical Operators
| Op | Meaning |
|---|---|
| `and` | True only if **both** are True |
| `or` | True if **at least one** is True |
| `not` | flips True ↔ False |

**Truth table**

| A | B | A and B | A or B |
|---|---|---|---|
| T | T | T | T |
| T | F | F | T |
| F | T | F | T |
| F | F | F | F |

````python
age = 22
has_id = True
is_student = False

print(age >= 18 and has_id)            # True
print(is_student or age < 25)          # True
print(not has_id)                      # False

# Short-circuit evaluation: Python stops as soon as the answer is known
x = 0
print(x != 0 and 10 / x > 1)    # no ZeroDivisionError: right side never runs
````

| `remove(x)` | remove first occurrence |
| `pop(i)` | remove & return item (default last) |
| `index(x)` / `count(x)` | find / count |
| `sort()` / `reverse()` | in-place |

````python
tasks = ["code", "eat"]
tasks.append(["sleep", "dinner"])
print(tasks)
tasks.insert(1, "debug")
# tasks.extend(["repeat", "coffee"])
print(tasks)

tasks.remove("eat")
print(tasks)
last = tasks.pop()
print(last, tasks)

nums = [5, 2, 9, 1, 7]
print(sorted(nums))                # returns a NEW sorted list
print(sorted(nums, reverse=True))
nums.sort()                        # sorts IN PLACE (returns None!)
print(nums)

print(sum(nums), min(nums), max(nums), len(nums))
print(3 in nums, 7 in nums)
````

### Set operations (Venn diagrams in code)

````python
ml_students = {"Ali", "Sara", "Zain", "Hira"}
web_students = {"Sara", "Hira", "Omar"}

print("Both (intersection):", ml_students & web_students)
print("Either (union):     ", ml_students | web_students)
print("Only ML (difference):", ml_students - web_students)
print("In exactly one (symmetric diff):", ml_students ^ web_students)
print({"Ali"} <= ml_students)          # subset check
````

> **ML bridge:** `set(labels)` gives you the **unique classes** in a classification dataset; set operations find overlap between train/test IDs (**data leakage** checks!); `len(set(words))` is the **vocabulary size** in NLP.

### Summary: choosing the right collection
| | `list` | `tuple` | `dict` | `set` |
|---|---|---|---|---|
| Syntax | `[1,2]` | `(1,2)` | `{"a":1}` | `{1,2}` |
| Ordered | Yes | Yes | Yes (by insertion) | No |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Yes | Yes | keys No | No |
| Access by | index | index | key | membership only |


# Conditional Statements

Programs need to **make decisions**. Conditionals run code only when a condition is `True`.

```python
if condition:
    # runs if condition is True
elif other_condition:
    # runs if the first was False and this is True
else:
    # runs if none of the above were True
```
Note: the **colon `:`** and **indentation** (4 spaces) are mandatory.

````python
temperature = 32

if temperature > 35:
    print("Extremely hot")
elif temperature > 25:
    print("Warm")
elif temperature > 15:
    print("Pleasant")
else:
    print("Cold")

print("This line always runs (not indented).")
````

### Flow of control
Python checks conditions **top to bottom** and runs **only the first** block that is `True`, then skips the rest.
Order matters!

````python
marks = 85

# Wrong order: always prints "Pass" for 85 because 85 >= 50 is checked first
if marks >= 50:
    print("Pass")
elif marks >= 80:
    print("Distinction")

# Correct order: most specific first
if marks >= 80:
    print("Distinction")
elif marks >= 50:
    print("Pass")
else:
    print("Fail")
````

### Combining conditions & nesting

````python
age = 20
has_license = True

if age >= 18 and has_license:
    print("You may drive")

# Nested
username, password = "admin", "1234"
if username == "admin":
    if password == "1234":
        print("Welcome!")
    else:
        print("Wrong password")
else:
    print("Unknown user")

# Truthiness makes code shorter
items = []
if not items:
    print("Cart is empty")
````

### One-line conditional (ternary expression)
`value_if_true if condition else value_if_false`

````python
age = 17
status = "adult" if age >= 18 else "minor"
print(status)

nums = [3, -1, 4, -5]
abs_like = [n if n >= 0 else -n for n in nums]
print(abs_like)
````

## `break` and `continue`
- `break`: exit the loop **immediately**
- `continue`: skip to the **next iteration**
- `else` on a loop: runs only if the loop finished **without** `break`

````python
# break: stop at the first negative number
for n in [4, 8, -1, 6]:
    if n < 0:
        print("Found negative:", n)
        break
    print("OK:", n)

# continue: skip odd numbers
for n in range(1, 8):
    if n % 2 != 0:
        continue
    print("even:", n)

# for-else: prime check
n = 29
for d in range(2, int(n ** 0.5) + 1):
    if n % d == 0:
        print(n, "is not prime")
        break
else:
    print(n, "is prime")
````

2. Given `data = [4, 8, 15, 16, 23, 42]`, compute the **mean** and **variance** using only loops (variance = average of squared differences from the mean).

````python
# Your code here
````

````python
#@title Solution { display-mode: "form" }
for n in range(1, 21):
    if n % 15 == 0: print("FizzBuzz")
    elif n % 3 == 0: print("Fizz")
    elif n % 5 == 0: print("Buzz")
    else: print(n)

data = [4, 8, 15, 16, 23, 42]
total = 0
for x in data:
    total += x
mean = total / len(data)

sq_diff_sum = 0
for x in data:
    sq_diff_sum += (x - mean) ** 2
variance = sq_diff_sum / len(data)
print("Mean:", round(mean, 2), "Variance:", round(variance, 2))
````

### Mini-game: Number guessing (with a simple strategy)

````python
import random
random.seed(7)                       # fixed seed -> same results every run
secret = random.randint(1, 100)

low, high, tries = 1, 100, 0
while True:
    guess = (low + high) // 2        # binary search strategy
    tries += 1
    if guess == secret:
        print(f"Found {secret} in {tries} tries!")
        break
    elif guess < secret:
        low = guess + 1
    else:
        high = guess - 1
````

## Variable scope and lifetime
Python looks up names using **LEGB**: **L**ocal → **E**nclosing → **G**lobal → **B**uilt-in.

- A **local** variable exists only while the function runs (its *lifetime*).
- A **global** variable lives for the whole program.

````python
counter = 0                      # global

def show():
    message = "I'm local"        # local: dies when function ends
    print(message, "| counter is", counter)   # can READ globals

show()

def increment():
    global counter               # explicitly modify the global
    counter += 1

increment(); increment()
print("counter:", counter)

# Better design: pass values in, return values out
def increment_pure(c):
    return c + 1
counter = increment_pure(counter)
print(counter)
````

````python
# Shadowing: a local with the same name hides the global
x = "global"
def f():
    x = "local"
    print("inside:", x)
f()
print("outside:", x)
````

### Gotcha: never use a mutable default argument

````python
def bad_append(item, bucket=[]):      # the SAME list is reused on every call!
    bucket.append(item)
    return bucket

print(bad_append(1))
print(bad_append(2))                  # [1, 2]: surprise!

def good_append(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print(good_append(1))
print(good_append(2))                 # [2]
````

Then write `normalize(values)` that returns each value as `(x - mean) / std`. This is **standardization**, a core ML preprocessing step. Test with `[10, 20, 30, 40, 50]`.

````python
# Your code here
````

````python
#@title Solution { display-mode: "form" }
def mean(values):
    """Return the average of a list of numbers."""
    return sum(values) / len(values)

def variance(values):
    """Return the population variance."""
    m = mean(values)
    return sum((x - m) ** 2 for x in values) / len(values)

def std_dev(values):
    """Return the population standard deviation."""
    return variance(values) ** 0.5

def normalize(values):
    """Standardize values to mean 0 and std 1."""
    m, s = mean(values), std_dev(values)
    return [round((x - m) / s, 3) for x in values]

data = [10, 20, 30, 40, 50]
print(mean(data), variance(data), round(std_dev(data), 3))
print(normalize(data))
````

````python
#@title Reference Solution { display-mode: "form" }
def parse_record(line: str) -> dict:
    """Convert 'Name,Dept,s1,s2,s3' into a clean record or raise ValueError."""
    parts = [p.strip() for p in line.split(",")]
    name, dept, *score_strs = parts
    if not name:
        raise ValueError("missing name")
    if len(score_strs) != 3:
        raise ValueError(f"expected 3 scores, got {len(score_strs)}")
    try:
        scores = [float(s) for s in score_strs]
    except ValueError:
        raise ValueError("non-numeric score")
    return {"name": name.title(), "dept": dept, "scores": scores}

def average(scores):
    return sum(scores) / len(scores)

def grade(avg):
    if avg >= 85: return "A"
    elif avg >= 70: return "B"
    elif avg >= 55: return "C"
    return "F"

valid, rejected = [], []
for line in raw_records:
    try:
        valid.append(parse_record(line))
    except ValueError as e:
        rejected.append((line, str(e)))

for rec in valid:
    rec["avg"] = round(average(rec["scores"]), 1)
    rec["grade"] = grade(rec["avg"])

print(f"{'Name':<10}{'Dept':<6}{'Avg':>6}  Grade")
print("-" * 30)
for rec in sorted(valid, key=lambda r: r["avg"], reverse=True):
    print(f"{rec['name']:<10}{rec['dept']:<6}{rec['avg']:>6}  {rec['grade']}")

print("\nRejected rows:")
for line, reason in rejected:
    print(f"  {line!r:<25} -> {reason}")

dept_avgs = {}
for rec in valid:
    dept_avgs.setdefault(rec["dept"], []).append(rec["avg"])
print("\nDepartment summary:")
for dept, avgs in dept_avgs.items():
    print(f"  {dept}: {round(average(avgs), 1)} ({len(avgs)} students)")

print("\nDepartments:", {r["dept"] for r in valid})
top = max(valid, key=lambda r: r["avg"])
print("Top student:", top["name"], top["avg"])
````

````python
class Student:
    """A university student."""

    def __init__(self, name, roll_no, semester):    # constructor: runs on creation
        self.name = name                             # instance attributes
        self.roll_no = roll_no
        self.semester = semester
        self.marks = []

    def add_marks(self, score):                      # method
        self.marks.append(score)

    def average(self):
        return sum(self.marks) / len(self.marks) if self.marks else 0.0

    def __str__(self):                               # what print() shows
        return f"{self.name} (Roll {self.roll_no}, Sem {self.semester})"

s1 = Student("Ali", "AI-101", 4)
s2 = Student("Sara", "AI-102", 5)

s1.add_marks(85); s1.add_marks(91)
print(s1)
print(s1.average())
print(s2.average())            # each object keeps its OWN data
print(type(s1), isinstance(s1, Student))
````

## 14.3 `__str__`, `__repr__`, and other "dunder" (magic) methods
Special methods let your objects work with Python's built-in syntax.

````python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):                       # unambiguous, for developers
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):                 # enables v1 + v2
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, k):                     # enables v * 3
        return Vector(self.x * k, self.y * k)

    def __eq__(self, other):                  # enables v1 == v2
        return self.x == other.x and self.y == other.y

    def __len__(self):                        # enables len(v)
        return 2

    def magnitude(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5

v1, v2 = Vector(1, 2), Vector(3, 4)
print(v1 + v2, v1 * 3, v1 == Vector(1, 2), len(v1), v2.magnitude())
````

## 14.4 Encapsulation: protecting data
Python uses **conventions**: a single underscore `_x` means "internal, please don't touch"; double underscore `__x` triggers name-mangling. Use **`@property`** to control access and add validation.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance            # "protected" by convention

    @property
    def balance(self):                     # read-only access: acct.balance
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

acct = BankAccount("Hira", 1000)
acct.deposit(500)
print(acct.balance)

try:
    acct.withdraw(5000)
except ValueError as e:
    print("Error:", e)

try:
    acct.balance = 1_000_000               # property has no setter
except AttributeError as e:
    print("Cannot set balance directly:", e)
````

## 14.5 Inheritance: reuse and specialize
A **child class** inherits everything from a **parent class** and can add or **override** behaviour. `super()` calls the parent's version.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"Hi, I'm {self.name}"

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)             # reuse parent's setup
        self.subject = subject

    def introduce(self):                   # override
        return f"{super().introduce()} and I teach {self.subject}"

class Learner(Person):
    def __init__(self, name, bootcamp):
        super().__init__(name)
        self.bootcamp = bootcamp

people = [Person("Ahmed"), Teacher("Ms. Noor", "Python"), Learner("Zain", "AI Bootcamp")]
for p in people:                           # POLYMORPHISM: same call, different behaviour
    print(p.introduce())

print(isinstance(people[1], Person), issubclass(Teacher, Person))
````

````python
# Polymorphism with shapes
import math

class Shape:
    def area(self):
        raise NotImplementedError("Subclasses must implement area()")

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return round(math.pi * self.r ** 2, 2)

class Rectangle(Shape):
    def __init__(self, w, h): self.w, self.h = w, h
    def area(self): return self.w * self.h

for shape in [Circle(3), Rectangle(4, 5)]:
    print(type(shape).__name__, shape.area())
````

## 14.6 Composition ("has-a") and `@classmethod` / `@staticmethod`
Prefer **composition** when one object *contains* others (a `Classroom` *has* `Student`s). Don't force inheritance where "is-a" doesn't fit.

````python
class Classroom:
    def __init__(self, name):
        self.name = name
        self.students = []                 # composition: holds Student objects

    def enroll(self, student):
        self.students.append(student)

    def class_average(self):
        avgs = [s.average() for s in self.students if s.marks]
        return round(sum(avgs) / len(avgs), 2) if avgs else 0

    @classmethod
    def from_names(cls, name, student_names):      # alternative constructor
        room = cls(name)
        for i, n in enumerate(student_names, start=1):
            room.enroll(Student(n, f"R-{i:03}", 1))
        return room

    @staticmethod
    def is_passing(avg):                           # utility: no self/cls needed
        return avg >= 50

room = Classroom.from_names("Python-A", ["Ali", "Sara", "Zain"])
room.students[0].add_marks(80)
room.students[1].add_marks(65)
print(room.class_average(), Classroom.is_passing(room.class_average()))
````

## 14.8 ML bridge: how scikit-learn thinks
Every scikit-learn model is a **class** with `fit()` and `predict()` methods. Let's build one ourselves: simple linear regression `y = a*x + b`.

````python
class SimpleLinearRegression:
    """y = slope * x + intercept, fitted with the closed-form solution."""

    def __init__(self):
        self.slope = None
        self.intercept = None

    def fit(self, X, y):
        n = len(X)
        mean_x, mean_y = sum(X) / n, sum(y) / n
        num = sum((xi - mean_x) * (yi - mean_y) for xi, yi in zip(X, y))
        den = sum((xi - mean_x) ** 2 for xi in X)
        self.slope = num / den
        self.intercept = mean_y - self.slope * mean_x
        return self                                  # allows chaining

    def predict(self, X):
        if self.slope is None:
            raise RuntimeError("Call fit() before predict()")
        return [self.slope * xi + self.intercept for xi in X]

    def score(self, X, y):                           # R-squared
        preds = self.predict(X)
        mean_y = sum(y) / len(y)
        ss_res = sum((yi - pi) ** 2 for yi, pi in zip(y, preds))
        ss_tot = sum((yi - mean_y) ** 2 for yi in y)
        return 1 - ss_res / ss_tot

hours_studied = [1, 2, 3, 4, 5, 6]
exam_score    = [52, 58, 65, 71, 78, 83]

model = SimpleLinearRegression().fit(hours_studied, exam_score)
print(f"slope={model.slope:.2f}, intercept={model.intercept:.2f}")
print("Predict 7 hours:", [round(p, 1) for p in model.predict([7])])
print("R^2:", round(model.score(hours_studied, exam_score), 4))
````

scikit-learn's `LinearRegression()` has the same `fit` / `predict` / `score` interface.

````python
#@title Solution { display-mode: "form" }
class Employee:
    def __init__(self, name, base_salary):
        self.name, self.base_salary = name, base_salary
    def monthly_pay(self):
        return self.base_salary

class Manager(Employee):
    def monthly_pay(self):
        return self.base_salary * 1.2

class Intern(Employee):
    def monthly_pay(self):
        return self.base_salary * 0.5

staff = [Employee("Omar", 100000), Manager("Noor", 150000), Intern("Zain", 60000)]
for e in staff:
    print(f"{e.name:<6} {type(e).__name__:<9} {e.monthly_pay():,.0f}")
````
