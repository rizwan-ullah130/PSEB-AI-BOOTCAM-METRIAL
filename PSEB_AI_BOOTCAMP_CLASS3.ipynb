# MODULE 19: Git and GitHub

**Git** records snapshots of your project so you can go back in time, try ideas safely, and work with others. **GitHub** is a website that hosts Git repositories: a backup, a team workspace and your public portfolio.

```
 working folder --git add--> staging area --git commit--> local repository --git push--> GitHub
```

| Term | Meaning |
|---|---|
| Repository (repo) | project folder plus its full history |
| Commit | a saved snapshot with a message |
| Branch | an independent line of work (`main` is the default) |
| Merge | combine one branch into another |
| Remote (`origin`) | the copy of the repo on GitHub |
| Pull request (PR) | a request to merge your branch, with review and discussion |

**Undoing things (commands you will really use)**

| I want to... | Command |
|---|---|
| unstage a file | `git restore --staged file` |
| throw away uncommitted edits in a file | `git restore file` |
| fix the last commit message | `git commit --amend -m "new message"` |
| undo the last commit but keep the changes | `git reset --soft HEAD~1` |
| undo a commit that is already shared | `git revert <hash>` (adds a new commit that cancels it) |
| temporarily shelve work | `git stash` then `git stash pop` |

## 19.2 Branches and merging
Never develop directly on `main`. Create a **feature branch**, work there, and merge when it is ready, so `main` always works.

## 19.5 Pull requests: how teams ship code
A **pull request** asks teammates to review your branch before it reaches `main`. This is the daily workflow in almost every software team:

1. **Update main:** `git switch main` then `git pull`
2. **Branch:** `git switch -c feature/add-discount`
3. **Commit** small, focused changes
4. **Push the branch:** `git push -u origin feature/add-discount`
5. On GitHub click **Compare & pull request**. Write a title and a description (template below). Pick a reviewer.
6. **Review:** teammates comment on lines of code; you push more commits to the same branch to answer them.
7. **Merge** the pull request on GitHub (*Squash and merge* keeps `main` history tidy), then **delete the branch**.
8. Locally: `git switch main`, `git pull`, `git branch -d feature/add-discount`.

**Pull request description template**
```
## What
Adds a percentage discount to the invoice total.

## Why
Sales needs to offer coupons (ticket #42).

## How to test
Run `pytest`. Try an invoice of 1000 with a 10 percent discount: expect 900.

## Checklist
- [ ] Tests added or updated
- [ ] No secrets or large files committed
- [ ] README updated if behaviour changed
```

**Reviewing code:** be kind and specific, check correctness, naming and tests, not personal style. **Protect `main`:** in repository settings, require a pull request and one approval before merging.

**Contributing to someone else's project:** *fork* it on GitHub, `git clone` your fork, add the original as a second remote (`git remote add upstream <url>`), work on a branch, push to your fork, open a PR to the original.

# MODULE 20: Virtual Environments and pip

Different projects need different packages and different **versions** of them. Project A may need `requests 2.28`, project B a newer one. Installing everything into one place leads to conflicts ("it works on my machine"). A **virtual environment** is a private folder with its own Python packages for **one project**.

| Concept | Meaning |
|---|---|
| `venv` | built-in tool that creates an isolated environment |
| `pip` | installs packages from PyPI (the Python Package Index) |
| `requirements.txt` | the exact list of packages and versions of a project |
| `pip freeze` | prints every installed package with its exact version |

| Task | Windows | macOS / Linux |
|---|---|---|
| create | `python -m venv .venv` | `python3 -m venv .venv` |
| activate | `.venv\Scripts\activate` | `source .venv/bin/activate` |
| leave | `deactivate` | `deactivate` |

Colab is already a private virtual machine, so you do not *need* a `venv` here. We create one anyway to see exactly what happens on your own laptop. Inside the cell we call the environment's own `python` and `pip` directly instead of activating it.

**Everyday pip commands**

| Command | Does |
|---|---|
| `pip install package` | install the latest version |
| `pip install "package==2.31.0"` | exact version (reproducible) |
| `pip install "package>=2.0,<3.0"` | version range |
| `pip install -r requirements.txt` | install everything the project lists |
| `pip install --upgrade package` | update |
| `pip uninstall package` | remove |
| `pip list`, `pip show package` | what is installed, details about one package |
| `pip freeze > requirements.txt` | save the current versions |
| `python -m pip ...` | run pip for **that** Python (avoids "wrong pip" problems) |

**Good habits**
- One environment per project, created inside the project folder as `.venv`.
- Commit `requirements.txt`; **never commit `.venv`** (add it to `.gitignore`).
- Pin versions in projects you deploy so results are reproducible.
- In notebooks use `%pip install package` (installs into the running kernel). `!pip install` may target a different Python.

## 21.2 REST in one page
A request has a **method**, a **URL** (`base/path?query=value`), optional **headers** (for example authentication) and sometimes a JSON **body**.

| Method | Purpose | Example |
|---|---|---|
| GET | read | get today's weather |
| POST | create | place a new order |
| PUT / PATCH | replace / update | change a delivery address |
| DELETE | remove | cancel a booking |

| Status | Meaning |
|---|---|
| 200 / 201 | OK / Created |
| 400 | bad request (your input is wrong) |
| 401 / 403 | not logged in / not allowed |
| 404 | not found |
| 429 | too many requests: slow down |
| 500 and above | the server has a problem |

You can see exactly what `requests` will send **without** sending anything:

````python
import requests

prepared = requests.Request(
    "GET", "https://api.example.com/v1/orders",
    params={"status": "paid", "limit": 5},
    headers={"Authorization": "Bearer <token>", "Accept": "application/json"},
).prepare()
print(prepared.method, prepared.url)
print(prepared.headers)
````

## 21.3 Your first GET request
We use **JSONPlaceholder**, a free practice API that needs no key. Real networks fail, so the helper below always sets a **timeout**, checks the status with `raise_for_status()` and handles errors. The sample data keeps the lesson working offline.

````python
def fetch_json(url, params=None, headers=None, timeout=10):
    """GET a URL and return parsed JSON, or None if anything goes wrong."""
    try:
        response = requests.get(url, params=params, headers=headers, timeout=timeout)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.HTTPError as e:
        print("HTTP error:", e.response.status_code)
    except requests.exceptions.Timeout:
        print("Request timed out")
    except requests.exceptions.RequestException as e:
        print("Request failed:", type(e).__name__)
    except ValueError:
        print("Response was not valid JSON")
    return None

SAMPLE_POST = {"userId": 1, "id": 1, "title": "sample post title", "body": "sample body text"}

post = fetch_json("https://jsonplaceholder.typicode.com/posts/1") or SAMPLE_POST
print(post["id"], "|", post["title"])

r = None
try:
    r = requests.get("https://jsonplaceholder.typicode.com/posts/1", timeout=10)
    print(r.status_code, r.headers.get("Content-Type"), r.elapsed.total_seconds(), "seconds")
except requests.exceptions.RequestException:
    print("(offline: skipping the response details)")
````

## 21.5 Query parameters and pagination
APIs rarely return everything at once. You ask for **pages** and loop until you have enough.

````python
def fetch_all_posts(pages=3, per_page=5):
    posts = []
    for page in range(1, pages + 1):
        batch = fetch_json("https://jsonplaceholder.typicode.com/posts", params={"_page": page, "_limit": per_page})
        if not batch:                                     # empty page or failure: stop
            break
        posts.extend(batch)
    return posts

posts = fetch_all_posts() or [dict(SAMPLE_POST, id=i) for i in range(1, 16)]
print(len(posts), "posts collected; ids:", [p["id"] for p in posts])
````

## 21.7 Secrets, sessions, retries
- **API keys are passwords.** Keep them in Colab **Secrets** (key icon) or environment variables, never in code you commit. Send them in a header: `headers={"Authorization": f"Bearer {key}"}`.
- A `requests.Session` reuses the connection and can carry default headers for every call.
- On **429** (too many requests) or a temporary failure, wait and retry, doubling the wait each time (**exponential backoff**).

````python
import os, time

def get_secret(name):
    """Colab Secrets first, then environment variables. Returns None if missing."""
    try:
        from google.colab import userdata
        return userdata.get(name)
    except Exception:
        return os.environ.get(name)

api_key = get_secret("WEATHER_API_KEY")
print("Key found" if api_key else "No key set (add it in Colab Secrets; never paste it into a cell)")

session = requests.Session()
session.headers.update({"Accept": "application/json", "User-Agent": "bootcamp-day3"})   # sent with every call

def with_retries(func, attempts=4, base_delay=0.2):
    """Call func(); on failure wait 0.2s, 0.4s, 0.8s... and try again."""
    for attempt in range(1, attempts + 1):
        try:
            return func()
        except Exception as e:
            if attempt == attempts:
                raise
            delay = base_delay * 2 ** (attempt - 1)
            print(f"attempt {attempt} failed ({e}); retrying in {delay:.1f}s")
            time.sleep(delay)

calls = {"n": 0}
def flaky_service():
    calls["n"] += 1
    if calls["n"] < 3:
        raise ConnectionError("temporary failure")
    return {"status": "ok"}

print(with_retries(flaky_service))
````

````python
#@title Solution { display-mode: "form" }

_cache = {}

def cached_rates():
    if "rates" not in _cache:
        _cache["rates"] = get_rates("USD")          # network call happens only the first time
    return _cache["rates"]

def convert(amount, from_currency, to_currency):
    rates = cached_rates()
    for code_ in (from_currency, to_currency):
        if code_ not in rates:
            raise ValueError(f"Unknown currency: {code_}")
    return round(amount / rates[from_currency] * rates[to_currency], 2)

print(convert(100, "USD", "PKR"))
print(convert(50_000, "PKR", "SAR"))
try:
    convert(10, "USD", "XYZ")
except ValueError as e:
    print("Error:", e)
````

## 22.4 Debugging
**Three kinds of errors:** *syntax* (Python refuses to run), *runtime* (an exception), *logic* (it runs and gives the **wrong answer**, the hardest kind).

**Read a traceback from the bottom:** the last line says *what*, the lines above show *where* and *who called it*.
```
Traceback (most recent call last):
  File "report.py", line 14, in <module>
    total = parse_amount(row["amount"])
  File "report.py", line 6, in parse_amount
    return float(text)
ValueError: could not convert string to float: 'N/A'
```

**A method that always works**
1. **Reproduce** the bug reliably.
2. **Isolate** the smallest input that fails.
3. **Form a hypothesis**, then test it with a print, a log line, an `assert` or the debugger. Change **one thing at a time**.
4. **Fix it, then add a test** so it never returns.

### Tool 1: print and logging

````python
import logging

logging.basicConfig(level=logging.DEBUG, format="%(levelname)-7s | %(message)s", force=True)
log = logging.getLogger("billing")

def average_order(orders):
    log.debug("received %d orders", len(orders))
    paid = [o for o in orders if o.get("status") == "paid"]
    x = len(paid)
    print(f"{x = }")                                   # f"{x = }" prints name and value: a quick peek
    if not paid:
        log.warning("no paid orders, returning 0")
        return 0
    return sum(o["amount"] for o in paid) / len(paid)
    print("sum executed")

print(average_order([{"status": "paid", "amount": 900}, {"status": "refunded", "amount": 500}, {"status": "paid", "amount": 1100}]))
print(average_order([]))
logging.getLogger().setLevel(logging.WARNING)           # silence debug messages with one line
````

### Tool 2: the debugger (`pdb`)
`breakpoint()` (or `python -m pdb script.py`) pauses the program so you can inspect it: `n` next line, `s` step into, `p expr` print, `c` continue, `q` quit. In a notebook you can also run `%debug` in a new cell right after an exception.

A real-life bug: a courier needs 3 parcels for 25 items when each parcel holds 10, but the program says 2. We feed debugger commands automatically so you can see a session:

````python
%%bash
mkdir -p pdb_demo && cd pdb_demo
cat > parcels.py <<'EOF'
def parcels_needed(items, capacity):
    return items // capacity

print(parcels_needed(25, 10))
EOF
printf 'b 2\nc\np items\np capacity\np items // capacity\np -(-items // capacity)\nq\n' | python -m pdb parcels.py 2>&1 | tail -12
````

The last expression, `-(-items // capacity)`, is a ceiling division: `25 / 10` rounds **up** to 3. Fix: `math.ceil(items / capacity)`. Then lock it in with a test.

### Classic bugs
| Bug | Symptom | Fix |
|---|---|---|
| off-by-one | loop skips first or last item | check `range` bounds, test boundaries |
| `//` instead of ceiling | not enough parcels, pages, batches | `math.ceil` |
| float equality | `0.1 + 0.2 == 0.3` is False | `math.isclose`, round money |
| mutable default argument | data leaks between calls | default to `None` |
| editing a list while looping | items skipped | build a new list |
| string versus number | `"5" + 5` fails, `"10" < "9"` is True | convert types early |
| wrong variable scope | `UnboundLocalError` | pass values in, return results |

# MODULE 23: SQL Basics with SQLite

A **database** keeps data safely between runs and lets you ask questions about it. **SQL** is the language for relational databases (tables of rows and columns). Python includes **`sqlite3`**, a complete database in a single file, so nothing needs installing. The same SQL works in PostgreSQL and MySQL with small differences.

Our case: a personal **expense tracker**.

````python
import sqlite3
import pandas as pd

conn = sqlite3.connect("toolkit.db")                 # a real file this time (":memory:" would be temporary)
conn.executescript("""
DROP TABLE IF EXISTS expenses;
DROP TABLE IF EXISTS budgets;
CREATE TABLE expenses (
    id        INTEGER PRIMARY KEY,
    spent_on  TEXT    NOT NULL,                     -- ISO dates (YYYY-MM-DD) sort and compare correctly
    category  TEXT    NOT NULL,
    amount    REAL    NOT NULL CHECK (amount > 0),
    note      TEXT
);
CREATE TABLE budgets (
    category       TEXT PRIMARY KEY,
    monthly_limit  REAL NOT NULL
);
""")

expenses = [
    ("2026-10-28", "Food", 1850, "groceries"), ("2026-11-01", "Bills", 14200, "electricity"),
    ("2026-11-02", "Transport", 3200, "fuel"), ("2026-11-03", "Food", 2400, "restaurant"),
    ("2026-11-05", "Fun", 4500, "cinema and snacks"), ("2026-11-07", "Food", 6100, "groceries"),
    ("2026-11-08", "Bills", 3900, "internet"), ("2026-11-09", "Transport", 2800, "ride sharing"),
    ("2026-11-11", "Fun", 5200, "concert ticket"), ("2026-11-12", "Food", 3300, "groceries"),
    ("2026-11-13", "Transport", 3400, "fuel"), ("2026-11-13", "Fun", 1200, "books"),
]
conn.executemany("INSERT INTO expenses (spent_on, category, amount, note) VALUES (?, ?, ?, ?)", expenses)
conn.executemany("INSERT INTO budgets VALUES (?, ?)", [("Food", 12000), ("Transport", 8000), ("Bills", 20000), ("Fun", 8000)])
conn.commit()

def show(sql, params=()):
    """Run a query and print the result as a table."""
    print(pd.read_sql_query(sql, conn, params=params).to_string(index=False), "\n")

show("SELECT COUNT(*) AS rows_loaded FROM expenses")
````

## 23.4 Changing data, safely
`UPDATE` and `DELETE` **always** need a `WHERE`. Use `?` placeholders for values (never f-strings, which allow SQL injection). `with conn:` makes a transaction: commit on success, rollback on error.

````python
with conn:
    conn.execute("UPDATE expenses SET amount = ? WHERE spent_on = ? AND note = ?", (3500, "2026-11-13", "fuel"))   # correct one row
    conn.execute("DELETE FROM expenses WHERE spent_on < ?", ("2026-11-01",))               # drop last month's data

show("SELECT COUNT(*) AS rows_left, SUM(amount) AS total FROM expenses")

try:
    with conn:
        conn.execute("INSERT INTO expenses (spent_on, category, amount) VALUES ('2026-11-14', 'Food', 900)")
        conn.execute("INSERT INTO expenses (spent_on, category, amount) VALUES ('2026-11-14', 'Food', -50)")   # violates CHECK
except sqlite3.IntegrityError as e:
    print("Rolled back, nothing was saved:", e)

show("SELECT COUNT(*) AS rows_still FROM expenses")

user_text = "x' OR '1'='1"                                                  # an attacker's search text
unsafe = f"SELECT * FROM expenses WHERE category = '{user_text}'"            # NEVER do this
print("Unsafe query returns", len(conn.execute(unsafe).fetchall()), "rows (it leaked everything)")
print("Safe query returns", len(conn.execute("SELECT * FROM expenses WHERE category = ?", (user_text,)).fetchall()), "rows")
````

````python
%%writefile tracker/rates_api.py
"""Fetch exchange rates from a public REST API."""
import requests

URL = "https://open.er-api.com/v6/latest/{base}"
SAMPLE_RATES = {"USD": 1, "PKR": 278.5, "EUR": 0.92, "SAR": 3.75, "GBP": 0.79}   # used for offline runs


class RatesError(Exception):
    """Raised when rates cannot be fetched or understood."""


def fetch_rates(base="USD", timeout=10):
    """Return {currency: rate} for one unit of the base currency."""
    try:
        response = requests.get(URL.format(base=base), timeout=timeout)
        response.raise_for_status()
        return response.json()["rates"]
    except (requests.exceptions.RequestException, KeyError, ValueError) as error:
        raise RatesError(f"could not fetch rates for {base}: {type(error).__name__}") from error
````

````python
%%writefile tracker/store.py
"""SQLite storage for exchange rates. All SQL lives in this file."""
import sqlite3
from datetime import datetime, timezone


class RateStore:
    def __init__(self, path=":memory:"):
        self.conn = sqlite3.connect(path)
        self.conn.row_factory = sqlite3.Row
        with self.conn:
            self.conn.execute(
                """CREATE TABLE IF NOT EXISTS rates (
                       id INTEGER PRIMARY KEY,
                       base TEXT NOT NULL,
                       currency TEXT NOT NULL,
                       rate REAL NOT NULL CHECK (rate > 0),
                       fetched_at TEXT NOT NULL)"""
            )

    def save(self, base, rates, currencies):
        """Store the chosen currencies from one fetch; return how many rows were saved."""
        missing = [c for c in currencies if c not in rates]
        if missing:
            raise ValueError(f"Unknown currency: {', '.join(missing)}")
        now = datetime.now(timezone.utc).isoformat(timespec="seconds")
        rows = [(base, c, rates[c], now) for c in currencies]
        with self.conn:
            self.conn.executemany(
                "INSERT INTO rates (base, currency, rate, fetched_at) VALUES (?, ?, ?, ?)", rows)
        return len(rows)

    def latest(self, base, currency):
        row = self.conn.execute(
            "SELECT rate FROM rates WHERE base = ? AND currency = ? ORDER BY id DESC LIMIT 1",
            (base, currency)).fetchone()
        return row["rate"] if row else None

    def history(self, base, currency, limit=5):
        rows = self.conn.execute(
            "SELECT fetched_at, rate FROM rates WHERE base = ? AND currency = ? ORDER BY id DESC LIMIT ?",
            (base, currency, limit))
        return [dict(r) for r in rows]

    def close(self):
        self.conn.close()
````

````python
%%writefile tracker/main.py
"""Currency tracker: fetch rates, store them, report the change since the last run.

Usage:
    python main.py            # live rates (falls back to sample data if the API fails)
    python main.py --offline  # always use sample data
"""
import sys

from rates_api import RatesError, SAMPLE_RATES, fetch_rates
from store import RateStore

BASE = "USD"
TARGETS = ["PKR", "EUR", "SAR", "GBP"]


def build_report(store, rates):
    """Return a text table with each rate and its change since the previous saved rate."""
    lines = [f"{'Pair':<10}{'Rate':>10}{'Change':>10}"]
    for currency in TARGETS:
        previous = store.latest(BASE, currency)
        change = "" if previous is None else f"{(rates[currency] - previous) / previous * 100:+.2f}%"
        lines.append(f"{BASE + '/' + currency:<10}{rates[currency]:>10.3f}{change:>10}")
    return "\n".join(lines)


def main(argv):
    try:
        rates = SAMPLE_RATES if "--offline" in argv else fetch_rates(BASE)
    except RatesError as error:
        print(f"[warn] {error}; using sample rates")
        rates = SAMPLE_RATES
    store = RateStore("rates.db")
    print(build_report(store, rates))
    store.save(BASE, rates, TARGETS)
    store.close()


if __name__ == "__main__":
    main(sys.argv[1:])
````

````python
%%writefile tracker/tests/test_rates_api.py
import sys, pathlib
sys.path.insert(0, str(pathlib.Path(__file__).resolve().parents[1]))

import pytest
import requests
from rates_api import RatesError, fetch_rates


class FakeResponse:
    def __init__(self, payload=None, error=None):
        self.payload, self.error = payload, error

    def raise_for_status(self):
        if self.error:
            raise self.error

    def json(self):
        if self.payload is None:
            raise ValueError("no JSON")
        return self.payload


def patch_get(monkeypatch, response):
    monkeypatch.setattr(requests, "get", lambda *args, **kwargs: response)


def test_success_returns_rates(monkeypatch):
    patch_get(monkeypatch, FakeResponse({"result": "success", "rates": {"PKR": 278.5}}))
    assert fetch_rates("USD") == {"PKR": 278.5}


def test_http_error_becomes_rates_error(monkeypatch):
    patch_get(monkeypatch, FakeResponse(error=requests.exceptions.HTTPError("500")))
    with pytest.raises(RatesError):
        fetch_rates("USD")


def test_invalid_json_becomes_rates_error(monkeypatch):
    patch_get(monkeypatch, FakeResponse(payload=None))
    with pytest.raises(RatesError):
        fetch_rates("USD")


def test_missing_rates_key_becomes_rates_error(monkeypatch):
    patch_get(monkeypatch, FakeResponse({"result": "error"}))
    with pytest.raises(RatesError):
        fetch_rates("USD")
````
