````python
# Calculate Total => price, quantity
def calculate_total(price, quantity, discount_percent=0, tax_rate=0.17):
    """Return the final amount of an order."""
    subtotal = price * quantity
    discounted = subtotal * (1 - discount_percent / 100)
    return round(discounted * (1 + tax_rate), 2)

help(calculate_total)
# Default Parameters
print(calculate_total(1500, 2))                          # defaults: no discount, 17% tax
# Positional Arguments
print(calculate_total(1500, 2, 0.05))                      # 10% discount (positional)
# Keyword Arguments
print(calculate_total(price=1500, quantity=2, tax_rate=0.05))   # keyword arguments, any order
````

- **Parameters** are the names in the definition; **arguments** are the values you pass.
- Parameters with a **default** (`discount_percent=0`) are optional.
- `return` sends a value back **and exits**. A function without `return` gives back `None`. `print` shows a value to a person; `return` gives it to the program.

````python
#Calculate Order's Summary
def order_summary(prices):
    """Return the smallest, largest and average price (a tuple)."""
    return min(prices), max(prices), round(sum(prices) / len(prices), 2)

low, high, avg = order_summary([450, 1200, 800, 2999])
print(low, high, avg)

def shipping_cost(weight_kg, express=False):
    """Rs. 150 for the first kg, Rs. 80 per extra kg. Express doubles the price."""
    if weight_kg <= 0:
        return None                      # early return for invalid input
    cost = 150 + max(weight_kg - 1, 0) * 80
    return cost * 2 if express else cost

print(shipping_cost(0.5), shipping_cost(3), shipping_cost(3, express=True), shipping_cost(-2))
````

````python
# *args collects extra positional arguments (a tuple); **kwargs collects extra keyword arguments (a dict)
def average(*scores):
    return sum(scores) / len(scores)

def create_user(name, email, **extra):
    user = {"name": name, "email": email}
    user.update(extra)                   # any additional fields: city, role, ...
    return user

print(average(70, 85, 90))
print(create_user("Ayesha", "ayesha@example.com", city="Islamabad", role="analyst"))
````

## Scope: where a variable lives
Python looks up names in this order: **L**ocal, **E**nclosing, **G**lobal, **B**uilt-in (LEGB). A local variable disappears when the function ends. Prefer **passing values in and returning results** over changing globals.

````python
tax_rate = 0.17                          # global

def price_with_tax(price):
    tax = price * tax_rate               # reading a global is fine
    return price + tax                   # 'tax' is local: gone after the function ends

print(price_with_tax(1000))
# print(tax)  # NameError: local variables are not visible outside

def apply_tax_bad(price):
    global tax_rate                      # avoid: hidden side effects make bugs hard to find
    tax_rate = 0.5
    return price * (1 + tax_rate)
x = apply_tax_bad(1000)
print(x)
````

## Functions are values: `lambda`, `sorted(key=...)`, `map`, `filter`
You will see these constantly in data code (Pandas `apply`, sorting records).

````python
orders = [
    {"id": 101, "customer": "Asad",   "total": 4500},
    {"id": 102, "customer": "Sara",  "total": 12800},
    {"id": 103, "customer": "Usman", "total": 760},
    {"id": 104, "customer": "Hira",  "total": 9300},
]

by_value = sorted(orders, key=lambda o: o["total"], reverse=True)       # sort records by a field
print([o["customer"] for o in by_value])

big_orders = list(filter(lambda o: o["total"] >= 5000, orders))        # keep matching items
print([o["id"] for o in big_orders])

with_tax = list(map(lambda o: round(o["total"] * 1.17, 2), orders))    # transform every item
print(with_tax)
````

### Exercise: Electricity Bill Calculator

Write `electricity_bill(units, fixed_charge=200, tax_rate=0.15)` that returns a **dictionary** with `energy_cost`, `fixed_charge`, `tax` and `total`.
Pricing: first 100 units at Rs. 10, next 100 units at Rs. 15, every unit above 200 at Rs. 22. Tax applies to energy cost plus fixed charge.
Test with 80, 150 and 320 units.

````python
# Your code here
````

````python
#@title Solution { display-mode: "form" }

def electricity_bill(units, fixed_charge=200, tax_rate=0.15):
    """Return a bill breakdown using tiered pricing."""
    if units < 0:
        raise ValueError("units cannot be negative")
    tier1 = min(units, 100) * 10
    tier2 = min(max(units - 100, 0), 100) * 15
    tier3 = max(units - 200, 0) * 22
    energy = tier1 + tier2 + tier3
    tax = round((energy + fixed_charge) * tax_rate, 2)
    return {"energy_cost": energy, "fixed_charge": fixed_charge, "tax": tax,
            "total": round(energy + fixed_charge + tax, 2)}

for u in (80, 150, 320):
    print(u, electricity_bill(u))
````

````python
def read_first_line(path):
    try:
        f = open(path)
    except FileNotFoundError:
        print(f"File not found: {path}")
        return None
    else:                                    # no error: safe to continue
        with f:
            return f.readline().strip()
    finally:
        print("(finally always runs: good for cleanup)")

print(read_first_line("does_not_exist.csv"))
````

**Rules of thumb:** catch **specific** exceptions (never a bare `except:`), keep `try` blocks small, and never silence an error with `pass`. Raise your own errors to reject bad input early.

````python
def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("Amount must be positive")
    if amount > balance:
        raise ValueError(f"Insufficient funds: balance {balance}, requested {amount}")
    return balance - amount


def safe_float(value, default=None):
    """Convert to float, or return default. Handy when cleaning data."""
    try:
        return float(value)
    except (TypeError, ValueError):
        return default

print([safe_float(v) for v in ["12.5", "abc", None, 7]])
````

````python


class BankAccount:
    """A simple bank account that protects its balance."""

    def __init__(self, owner, balance=0):          # constructor: runs on BankAccount(...)
        self.owner = owner                         # instance attributes
        self.balance = balance                    # leading underscore: "internal, don't touch"
        self.history = []


    def balance(self):                             # read-only attribute: account.balance
        return self.balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount
        self.history.append(("deposit", amount))

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        self.history.append(("withdraw", amount))

    def __str__(self):                             # what print() shows
        return f"{self.owner}: Rs. {self.balance:,}"

acct = BankAccount("Hira", 10_000)
# acct.deposit(2_500)
print(acct.withdraw(4_000))
print(acct)
print(acct.history)
````

- `self` is the object itself; Python passes it automatically as the first parameter of every method.
- **Encapsulation**: keep data private (`_balance`) and expose safe methods (`deposit`, `withdraw`) that enforce the rules.

## 14.2 Inheritance and polymorphism: payment methods
A **child class** inherits from a **parent** and can **override** behaviour. **Polymorphism**: the same call (`fee()`) behaves differently per class.

````python
class Payment:
    name = "Generic"
    def fee(self, amount):
        return 0
    def total(self, amount):
        return round(amount + self.fee(amount), 2)

class CardPayment(Payment):
    name = "Card"
    def fee(self, amount):
        return amount * 0.025              # 2.5% processing fee

class WalletPayment(Payment):
    name = "Mobile wallet"
    def fee(self, amount):
        return amount * 0.01

class CashOnDelivery(Payment):
    name = "Cash on delivery"
    def fee(self, amount):
        return 100                         # flat handling fee

order_amount = 8_000
for method in (CardPayment(), WalletPayment(), CashOnDelivery()):
    print(f"{method.name:<18} fee={method.fee(order_amount):>7.2f}  total={method.total(order_amount):>9.2f}")

print(isinstance(CardPayment(), Payment))
````

## 14.4 How ML libraries use classes
Every scikit-learn model is a class with `fit()`, `predict()` and `score()`. Here is a tiny one: a baseline that always predicts the most common class (useful in EDA to judge whether a model is any good).

````python
from collections import Counter

class MajorityClassifier:
    """Baseline model: always predict the most frequent label seen in training."""

    def fit(self, y):
        self.majority_ = Counter(y).most_common(1)[0][0]    # trailing underscore: learned attribute
        return self

    def predict(self, n):
        return [self.majority_] * n

    def score(self, y):
        predictions = self.predict(len(y))
        return sum(p == t for p, t in zip(predictions, y)) / len(y)

labels = ["stay"] * 74 + ["leave"] * 26
model = MajorityClassifier().fit(labels)
print(model.majority_, "| accuracy of the do-nothing model:", model.score(labels))
````

Build an `Inventory` class that manages `Product` objects (reuse the dataclass above):
1. `add_product(product)`: store by name; raise `ValueError` if it already exists.
2. `sell(name, qty)`: reduce stock; raise `OutOfStockError` if there is not enough, `KeyError` if the product is unknown.
3. `low_stock(threshold=5)`: return the names of products at or below the threshold.
4. `total_value()`: sum of price * stock.
Test with three products, one successful sale, one failed sale, and a low-stock report.

````python
# Your code here
````

````python
#@title Solution { display-mode: "form" }

class Inventory:
    def __init__(self):
        self.products = {}

    def add_product(self, product):
        if product.name in self.products:
            raise ValueError(f"{product.name} already exists")
        self.products[product.name] = product

    def sell(self, name, qty):
        product = self.products[name]                  # KeyError if unknown
        if qty > product.stock:
            raise OutOfStockError(f"Only {product.stock} x {name} left")
        product.stock -= qty

    def low_stock(self, threshold=5):
        return [p.name for p in self.products.values() if p.stock <= threshold]

    def total_value(self):
        return sum(p.price * p.stock for p in self.products.values())

inv = Inventory()
for p in (Product("Keyboard", 3500, 12), Product("Webcam", 6500, 4), Product("Headset", 5200, 2)):
    inv.add_product(p)

inv.sell("Keyboard", 3)
try:
    inv.sell("Headset", 5)
except OutOfStockError as e:
    print("Sale failed:", e)
print("Low stock:", inv.low_stock())
print("Inventory value:", inv.total_value())
````

# Exploratory Data Analysis (EDA)

**EDA** is the investigation you do **before** modelling: understanding what the data contains, what is wrong with it, and what it can tell you. Models trained on unexamined data learn the data's mistakes ("garbage in, garbage out").

| Stage | Question | Part |
|---|---|---|
| Load and describe | What do I have? | 1 |
| Data quality audit | What is wrong with it? | 1 |
| Single columns | What does each variable look like? | 2 |
| Target and relationships | What predicts what I want to predict? | 2 |
| Clean | Fix missing values, outliers, text, types | 3 |
| Encode and engineer | Make it numeric and informative | 3 |
| Hand over | A clean dataset ready for modelling | 3 |

````python
def make_churn_data(n=2000, seed=42):
    """Simulate a messy telecom customer file (the kind a real company exports)."""
    rng = np.random.default_rng(seed)
    tenure = rng.integers(0, 73, n)
    contract = rng.choice(["Month-to-month", "One year", "Two year"], n, p=[.55, .25, .20])
    internet = rng.choice(["DSL", "Fiber optic", "No"], n, p=[.35, .45, .20])
    base = np.where(internet == "Fiber optic", 85, np.where(internet == "DSL", 55, 25))
    monthly = np.round(base + rng.normal(0, 8, n), 2)
    calls = rng.poisson(1.5, n)
    payment = rng.choice(["Electronic check", "Mailed check", "Bank transfer", "Credit card"], n, p=[.35, .2, .25, .2])
    z = (-1.5 + 1.3 * (contract == "Month-to-month") - 1.0 * (contract == "Two year")
         + 0.7 * (internet == "Fiber optic") - 0.035 * tenure + 0.012 * (monthly - 60)
         + 0.25 * calls + 0.4 * (payment == "Electronic check"))
    churn = rng.random(n) < 1 / (1 + np.exp(-z))
    df = pd.DataFrame({
        "customer_id": [f"C{1000 + i}" for i in range(n)],
        "gender": rng.choice(["Male", "Female"], n),
        "senior_citizen": rng.choice([0, 1], n, p=[.84, .16]),
        "partner": rng.choice(["Yes", "No"], n),
        "dependents": rng.choice(["Yes", "No"], n, p=[.3, .7]),
        "tenure": tenure,
        "tenure_years": np.round(tenure / 12, 2),
        "signup_date": (pd.Timestamp("2026-10-31") - pd.to_timedelta(tenure * 30, unit="D")).strftime("%Y-%m-%d"),
        "internet_service": internet,
        "contract": contract,
        "payment_method": payment,
        "paperless_billing": rng.choice(["Yes", "No"], n, p=[.6, .4]),
        "monthly_charges": monthly,
        "total_charges": np.round(monthly * tenure + rng.normal(0, 15, n), 2).clip(min=0).astype(str),
        "support_calls": calls.astype(float),
        "churn": np.where(churn, "Yes", "No"),
    })
    df.loc[df["tenure"] == 0, "total_charges"] = " "              # blank text, a classic export problem
    idx = lambda k: rng.choice(n, k, replace=False)
    df.loc[idx(80), "support_calls"] = np.nan                       # missing values
    df.loc[idx(60), "payment_method"] = np.nan
    df.loc[idx(40), "monthly_charges"] = np.nan
    df.loc[idx(5), "monthly_charges"] = 999.99                      # data-entry errors
    df.loc[idx(4), "support_calls"] = 40                            # extreme but real
    bad = idx(3); df.loc[bad, "tenure"] = -1; df.loc[bad, "tenure_years"] = -0.08   # impossible values
    messy = idx(60)
    df.loc[messy[:30], "contract"] = "month-to-month"               # inconsistent text
    df.loc[messy[30:], "contract"] = "Month-to-month "
    df.loc[idx(25), "gender"] = "M"
    df = pd.concat([df, df.sample(25, random_state=1)], ignore_index=True)   # duplicate rows
    return df

make_churn_data().to_csv("customer_churn_raw.csv", index=False)     # "export" the file
print("customer_churn_raw.csv created")
````

### 1.2 Audit data quality (before any chart)
Charts built on dirty data mislead. Check these **six** things first:

````python
# 1. Missing values
missing = df.isna().sum()
print(pd.DataFrame({"missing": missing, "percent": (missing / len(df) * 100).round(1)}).query("missing > 0"))

# 2. Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 3. Wrong data types: numbers stored as text
as_number = pd.to_numeric(df["total_charges"], errors="coerce")
print("total_charges values that are not numbers:", as_number.isna().sum())
print(df.loc[as_number.isna(), "total_charges"].head(3).tolist())
````

````python
# 4. Inconsistent categories: the same thing spelled differently
print(df["contract"].value_counts(), "\n")
print(df["gender"].value_counts(), "\n")

# 5. Impossible or suspicious values
print("Negative tenure:", (df["tenure"] < 0).sum())
print("Monthly charges above 200:", (df["monthly_charges"] > 200).sum())
print("Support calls above 20:", (df["support_calls"] > 20).sum())

# 6. Useless columns: identifiers (unique per row) or constants
for col in df.columns:
    if df[col].nunique() == len(df):
        print("Identifier column (drop before modelling):", col)
````

Turn the checks into a **reusable function**, because you will audit every dataset you ever receive:

````python
def data_quality_report(data):
    """Return one row per column: type, missing count and percent, unique values, example."""
    report = pd.DataFrame({
        "dtype": data.dtypes.astype(str),
        "missing": data.isna().sum(),
        "missing_%": (data.isna().mean() * 100).round(1),
        "unique": data.nunique(),
        "example": data.iloc[0],
    })
    print(f"Rows: {len(data):,} | Columns: {data.shape[1]} | Duplicate rows: {data.duplicated().sum()}")
    return report

data_quality_report(df)
````

### 2.1 One numeric column: centre, spread, shape, outliers
| Measure | Question it answers |
|---|---|
| mean, median | what is typical? (large gap between them = skewed or outliers) |
| std, IQR | how spread out? |
| skew | symmetric or long tail? |
| min / max / IQR rule | outliers? |

**IQR rule:** a value is an outlier if it is below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`.

````python
def describe_column(s):
    """Centre, spread, shape and outliers of one numeric column."""
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return pd.Series({
        "mean": s.mean(), "median": s.median(), "std": s.std(), "skew": s.skew(),
        "min": s.min(), "max": s.max(), "IQR": iqr,
        "lower_fence": low, "upper_fence": high,
        "outliers": int(((s < low) | (s > high)).sum()),
    }).round(2)

summary = pd.DataFrame({c: describe_column(eda[c]) for c in ["tenure", "monthly_charges", "total_charges", "support_calls"]})
summary
````

````python
def plot_distribution(s, title):
    """Histogram (shape) above a boxplot (outliers)."""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7, 4.5), sharex=True, gridspec_kw={"height_ratios": [3, 1]})
    sns.histplot(s.dropna(), bins=40, kde=True, ax=ax1)
    sns.boxplot(x=s.dropna(), ax=ax2)
    ax1.set_title(title)
    plt.tight_layout()
    plt.show()

plot_distribution(eda["monthly_charges"], "Monthly charges: the 999.99 values stretch the axis")
plot_distribution(eda["support_calls"], "Support calls: right-skewed (most customers call rarely)")
````

**Reading the plots:** `monthly_charges` has a normal body and a few extreme points (data entry errors: nobody pays 999.99). `support_calls` is **right-skewed**: mean above median, long tail. Different problems, different treatment (Part 3).

### 2.5 Correlation: strength of numeric relationships and redundant columns
Correlation ranges from -1 to +1. **Near 0** means no linear relationship. **Close to ±1 between two features** means they carry the same information: **redundant** (one can be dropped). Correlation is not causation, and it only measures *linear* relationships between numbers.

````python
numeric = eda.select_dtypes("number")
corr = numeric.corr()

print(corr["churned"].drop("churned").sort_values().round(2), "\n")

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation matrix")
plt.show()

def high_correlations(corr_matrix, threshold=0.75):
    """List feature pairs that are almost the same information."""
    pairs = []
    cols = corr_matrix.columns
    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            if abs(corr_matrix.iloc[i, j]) >= threshold:
                pairs.append((cols[i], cols[j], round(corr_matrix.iloc[i, j], 2)))
    return pairs

print("Redundant pairs:", high_correlations(corr))
````

**Findings:**
- `tenure` and `tenure_years` are the same measurement in different units (correlation 1.0), so one of them goes.
- `total_charges` moves with tenure (about 0.78) because it accumulates monthly charges: related but not identical.
- The strongest *numeric* link with churn is tenure (about -0.2): newer customers leave more often.
- Numeric correlations with churn are modest, yet the charts above show churn differs enormously by contract type. Correlation only measures linear relationships between numbers, so combine it with group comparisons for categorical features.

### 3.1 Duplicates, inconsistent text, wrong types, impossible values

````python
clean = df.copy()
print("Start:", clean.shape)

# 1. Duplicates: keep the first occurrence of identical rows
clean = clean.drop_duplicates()
print("After removing duplicates:", clean.shape)

# 2. Inconsistent text: strip spaces, lowercase, then map to one standard spelling
clean["contract"] = (clean["contract"].str.strip().str.lower()
                     .map({"month-to-month": "Month-to-month", "one year": "One year", "two year": "Two year"}))
clean["gender"] = clean["gender"].str.strip().str.lower().map({"male": "Male", "m": "Male", "female": "Female"})
print(clean["contract"].value_counts().to_dict(), clean["gender"].value_counts().to_dict())

# 3. Wrong type: convert text to numbers; unparseable values (blanks) become NaN
clean["total_charges"] = pd.to_numeric(clean["total_charges"], errors="coerce")

# 4. Impossible values: a negative tenure is wrong. Recompute it from the signup date instead of guessing
signup = pd.to_datetime(clean["signup_date"])
tenure_from_date = (pd.Timestamp("2026-10-31") - signup).dt.days // 30
print("Negative tenures fixed:", (clean["tenure"] < 0).sum())
clean.loc[clean["tenure"] < 0, "tenure"] = tenure_from_date
print(clean["tenure"].min(), clean["tenure"].max())
````

### 3.2 Outliers: detect, then decide
Detecting an outlier is mechanical (IQR rule). **Deciding** what to do needs business knowledge:

| Situation | Action |
|---|---|
| Clearly an **error** (999.99 monthly charge, negative age) | treat as missing, then fill |
| **Real but extreme** (40 support calls) | keep, or **cap** (winsorise) so it does not dominate the model |
| Unsure | keep and compare results with and without |

Never delete outliers just because they are inconvenient: they may be the most interesting customers.

````python
q1, q3 = clean["monthly_charges"].quantile([0.25, 0.75])
upper = q3 + 1.5 * (q3 - q1)
print(f"monthly_charges upper fence: {upper:.1f}; values above it: {(clean['monthly_charges'] > upper).sum()}")

# Errors: impossible price -> mark as missing; filled in the next step
clean.loc[clean["monthly_charges"] > upper, "monthly_charges"] = np.nan

# Real but extreme: cap at the 99th percentile
cap = clean["support_calls"].quantile(0.99)
print("support_calls 99th percentile cap:", cap)
clean["support_calls"] = clean["support_calls"].clip(upper=cap)
print(clean["support_calls"].max())
````

### 3.3 Missing values: drop or fill?
| Situation | Recommended action |
|---|---|
| Column is **mostly empty** (above about 50%) and not critical | drop the column |
| **Few rows** (below about 5%) are missing and it is random | drop those rows |
| Numeric, moderate missing | fill with the **median** (robust to outliers), or a **group median** |
| Categorical | fill with the **mode** or an explicit `"Unknown"` category |
| Missingness itself is informative | add a `was_missing` **flag** column before filling |
| Value can be **calculated** from other columns | calculate it |

Never fill using the test set, and always fill **after** deciding what is an error (previous step).

````python
print(clean.isna().sum()[lambda s: s > 0], "\n")

# Group-aware fill: typical charge depends on the internet service, so use the group median
clean["monthly_charges"] = clean["monthly_charges"].fillna(
    clean.groupby("internet_service")["monthly_charges"].transform("median"))

# Calculable: total charges = tenure x monthly charges
clean["total_charges"] = clean["total_charges"].fillna((clean["tenure"] * clean["monthly_charges"]).round(2))

# Informative missingness: keep a flag, then fill the number with the median
clean["support_calls_missing"] = clean["support_calls"].isna().astype(int)
clean["support_calls"] = clean["support_calls"].fillna(clean["support_calls"].median())

# Categorical: an explicit category is more honest than guessing
clean["payment_method"] = clean["payment_method"].fillna("Unknown")

print("Missing values left:", clean.isna().sum().sum())
````

### 3.4 Feature engineering: create columns that carry more signal
Use domain knowledge to build features a model cannot easily discover alone.

````python
clean["tenure_group"] = pd.cut(clean["tenure"], bins=[-1, 12, 24, 48, 72], labels=["0-12", "13-24", "25-48", "49-72"])
clean["is_new_customer"] = (clean["tenure"] <= 6).astype(int)
clean["avg_monthly_spend"] = (clean["total_charges"] / clean["tenure"].clip(lower=1)).round(2)
clean["signup_month"] = pd.to_datetime(clean["signup_date"]).dt.month

print(clean[["tenure", "tenure_group", "is_new_customer", "avg_monthly_spend", "signup_month"]].head())
print(clean.groupby("tenure_group", observed=True).size())
````

### 3.5 Encode categorical columns
Models need numbers. Choose the encoding by the type of column:

| Type | Example | Encoding |
|---|---|---|
| **Binary** | Yes / No | map to 1 / 0 |
| **Ordinal** (has a natural order) | contract length | map to 0, 1, 2 |
| **Nominal** (no order) | payment method | **one-hot** (`get_dummies`), `drop_first=True` to avoid duplicate information |

Never give nominal categories codes like 1, 2, 3 (the model would think "Credit card" is greater than "Bank transfer").

````python
yes_no = {"Yes": 1, "No": 0}
for col in ["partner", "dependents", "paperless_billing", "churn"]:
    clean[col] = clean[col].map(yes_no)
clean["is_male"] = (clean["gender"] == "Male").astype(int)
clean["contract"] = clean["contract"].map({"Month-to-month": 0, "One year": 1, "Two year": 2})   # ordinal

# Drop redundant and identifier columns, then one-hot the nominal ones
clean = clean.drop(columns=["customer_id", "tenure_years", "signup_date", "gender"])
clean = pd.get_dummies(clean, columns=["internet_service", "payment_method", "tenure_group"], drop_first=True, dtype=int)

print(clean.shape)
clean.head()
````

### 3.6 Package everything into one reusable function and verify
A cleaning process you ran by hand once is not reproducible. Put it in **one function** (you can run it on next month's file), then **assert** that the result meets the contract.

````python
def clean_churn(raw):
    """Raw customer export -> numeric, complete, model-ready DataFrame."""
    d = raw.drop_duplicates().copy()
    d["contract"] = d["contract"].str.strip().str.lower().map(
        {"month-to-month": "Month-to-month", "one year": "One year", "two year": "Two year"})
    d["gender"] = d["gender"].str.strip().str.lower().map({"male": "Male", "m": "Male", "female": "Female"})
    d["total_charges"] = pd.to_numeric(d["total_charges"], errors="coerce")
    from_date = (pd.Timestamp("2026-10-31") - pd.to_datetime(d["signup_date"])).dt.days // 30
    d.loc[d["tenure"] < 0, "tenure"] = from_date

    q1, q3 = d["monthly_charges"].quantile([0.25, 0.75])
    d.loc[d["monthly_charges"] > q3 + 1.5 * (q3 - q1), "monthly_charges"] = np.nan
    d["support_calls"] = d["support_calls"].clip(upper=d["support_calls"].quantile(0.99))

    d["monthly_charges"] = d["monthly_charges"].fillna(d.groupby("internet_service")["monthly_charges"].transform("median"))
    d["total_charges"] = d["total_charges"].fillna((d["tenure"] * d["monthly_charges"]).round(2))
    d["support_calls_missing"] = d["support_calls"].isna().astype(int)
    d["support_calls"] = d["support_calls"].fillna(d["support_calls"].median())
    d["payment_method"] = d["payment_method"].fillna("Unknown")

    d["tenure_group"] = pd.cut(d["tenure"], bins=[-1, 12, 24, 48, 72], labels=["0-12", "13-24", "25-48", "49-72"])
    d["is_new_customer"] = (d["tenure"] <= 6).astype(int)
    d["avg_monthly_spend"] = (d["total_charges"] / d["tenure"].clip(lower=1)).round(2)
    d["signup_month"] = pd.to_datetime(d["signup_date"]).dt.month

    for col in ["partner", "dependents", "paperless_billing", "churn"]:
        d[col] = d[col].map({"Yes": 1, "No": 0})
    d["is_male"] = (d["gender"] == "Male").astype(int)
    d["contract"] = d["contract"].map({"Month-to-month": 0, "One year": 1, "Two year": 2})
    d = d.drop(columns=["customer_id", "tenure_years", "signup_date", "gender"])
    return pd.get_dummies(d, columns=["internet_service", "payment_method", "tenure_group"], drop_first=True, dtype=int)

model_ready = clean_churn(df)

# The contract of a clean dataset: every line below must hold
assert model_ready.isna().sum().sum() == 0, "missing values remain"
assert model_ready.select_dtypes(exclude="number").empty, "non-numeric columns remain"
assert not model_ready.duplicated().any(), "duplicates remain"
assert (model_ready["tenure"] >= 0).all(), "negative tenure remains"
assert set(model_ready["churn"].unique()) == {0, 1}, "target is not 0/1"
print("All checks passed.", model_ready.shape)
model_ready.to_csv("customer_churn_clean.csv", index=False)
````

### 3.7 Hand-over: split correctly and check the data is usable
Split **before** any learning, with `stratify` so the churn rate is the same in both sets. The model below is only a smoke test: it proves the data works. Building and judging models is the next class.

````python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score

X = model_ready.drop(columns="churn")
y = model_ready["churn"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
print(f"Churn rate  train: {y_train.mean():.1%}   test: {y_test.mean():.1%}")

model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_train, y_train)
pred = model.predict(X_test)

baseline = 1 - y_test.mean()
print(f"Baseline accuracy (always 'stay'): {baseline:.1%}")
print(f"Model accuracy: {accuracy_score(y_test, pred):.1%} | recall for churners: {recall_score(y_test, pred):.1%} (baseline recall: 0%)")
````

The model beats the do-nothing baseline and, more importantly, finds a share of the churners that the baseline misses completely (recall above zero). It is far from perfect: improving it, and judging it with imbalance-aware metrics such as recall and precision, is the job of the modelling classes.

## 18.1 Reading errors and debugging
A **traceback** is read from the bottom: the last line says *what* went wrong, the lines above show *where* and *who called it*.

```
Traceback (most recent call last):
  File "report.py", line 12, in <module>
    total = parse_price(row["price"])
  File "report.py", line 5, in parse_price
    return float(text)
ValueError: could not convert string to float: 'free'
```

Debugging method: **reproduce** the bug, **read** the message, **isolate** a small failing input, change **one thing at a time**, then **add a test** so it never returns.

````python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)-8s | %(message)s", force=True)
log = logging.getLogger("pipeline")

def load_orders(rows):
    log.info("Loading %d rows", len(rows))
    good = []
    for r in rows:
        if r.get("price") is None:
            log.warning("Skipping row without price: %s", r)      # visible in production logs; print() is not
            continue
        good.append(r)
    return good

orders_in = [{"id": 1, "price": 500}, {"id": 2}, {"id": 3, "price": 900}]
print(len(load_orders(orders_in)), "rows kept")

# assert: state what must be true; fails loudly close to the real problem
assert all("price" in r for r in load_orders(orders_in)), "an order without price slipped through"
````

## 18.4 Calling a web API
An API lets your code ask another service for data, usually as JSON. Always set a **timeout**, call `raise_for_status()`, and handle failure.

````python
import requests

def fetch_json(url, params=None, timeout=10):
    """GET a URL and return parsed JSON, or None if the request fails."""
    try:
        response = requests.get(url, params=params, timeout=timeout)
        response.raise_for_status()                    # turns 4xx / 5xx into an exception
        return response.json()
    except requests.exceptions.RequestException as e:
        print("Request failed:", type(e).__name__)
        return None

posts = fetch_json("https://jsonplaceholder.typicode.com/posts", params={"userId": 1})
if posts is None:                                       # offline fallback so the lesson still runs
    posts = [{"id": 1, "title": "sample post"}]
print(len(posts), "posts; first title:", posts[0]["title"])

posts_df = pd.DataFrame(posts)                          # API JSON -> DataFrame, ready for EDA
print(posts_df.shape)
````
